### Delete non-use keys in chunk_text

```python
keep_keys = ["condition", "section", "url", "chunk_index"]

for i in range(len(df["chunk_text"])):
    meta = df["chunk_text"][i]["meta"]
    cleaned_meta = {k: meta[k] for k in keep_keys if k in meta}
    df["chunk_text"][i]["meta"] = cleaned_meta
```

### Save dataframe

```python
df.to_json('../data/nhs_chunking_cleaned.jsonl', lines=True, orient='records', force_ascii=False)
```
